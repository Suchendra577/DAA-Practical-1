# DAA Practical 3: Heap Sort Algorithm

## Overview
**Heap Sort** is a comparison-based sorting technique based on a **Binary Heap** data structure. It is similar to Selection Sort where we first find the maximum element and place it at the end, repeating the process for the remaining elements.

### Key Phases of Heap Sort:
1. **Build a Max-Heap**: Transform the unsorted array into a Max-Heap where the root node `arr[0]` contains the maximum element.
2. **Extract Elements**: Repeatedly swap the root (maximum element) with the last element of the heap, reduce heap size by 1, and call `heapify` on the root node to maintain the Max-Heap property.

---

## Algorithm Implementation & Execution

```python
import time

def heapify(arr, n, i):
    """
    To heapify a subtree rooted with node i, which is an index in arr[].
    n is the size of the heap.
    """
    largest = i  # Initialize largest as root
    left = 2 * i + 1  # left child = 2*i + 1
    right = 2 * i + 2  # right child = 2*i + 2

    # See if left child of root exists and is greater than root
    if left < n and arr[left] > arr[largest]:
        largest = left

    # See if right child of root exists and is greater than current largest
    if right < n and arr[right] > arr[largest]:
        largest = right

    # Change root, if needed
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # swap

        # Heapify the root recursively
        heapify(arr, n, largest)

def heap_sort(arr):
    """
    Main function to perform Heap Sort on array arr.
    """
    n = len(arr)

    # Step 1: Build a Max-Heap. Start from last non-leaf node and heapify upwards.
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # Step 2: One by one extract elements from heap
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]  # Swap current root to end
        heapify(arr, i, 0)  # Call max heapify on the reduced heap

    return arr

# Get user input for the array
input_str = input("Enter numbers separated by spaces: ")
user_array = list(map(int, input_str.split()))

print(f"Original array: {user_array}")

start_time = time.time()
sorted_array = heap_sort(user_array)
end_time = time.time()
execution_time = end_time - start_time

print(f"Sorted array: {sorted_array}")
print(f"Execution time: {execution_time:.6f} seconds")
```

## Complexity Analysis

| Metric | Complexity | Explanation |
| :--- | :--- | :--- |
| **Best Case Time Complexity** | $\mathcal{O}(N \log N)$ | Occurs when all elements are distinct or reverse sorted. |
| **Average Case Time Complexity** | $\mathcal{O}(N \log N)$ | Heapifying takes $\mathcal{O}(\log N)$ for each of the $N$ elements. |
| **Worst Case Time Complexity** | $\mathcal{O}(N \log N)$ | Guaranteed upper bound regardless of initial array ordering. |
| **Auxiliary Space Complexity** | $\mathcal{O}(1)$ | In-place sorting algorithm requiring constant extra space. |
| **Stability** | Not Stable | Equal elements may lose their relative initial order during swaps. |
