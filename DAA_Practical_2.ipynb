```python
import time

def linear_search(arr, target):
    """
    Performs a linear search to find the target element in the given array.
    Returns the index of the target if found, otherwise returns -1.
    """
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# Get user input for the list
list_str = input("Enter a list of numbers separated by spaces (e.g., 10 20 30 40 50): ")
my_list = list(map(int, list_str.split()))

# Get user input for the target element
target_element = int(input("Enter the number to search for: "))

# Measure execution time
start_time = time.time()
index = linear_search(my_list, target_element)
end_time = time.time()

execution_time = (end_time - start_time) * 1000 # Convert to milliseconds

print(f"\nList: {my_list}")
print(f"Target Element: {target_element}")

if index != -1:
    print(f"Element {target_element} found at index {index}.")
else:
    print(f"Element {target_element} not found in the list.")

print(f"Execution Time: {execution_time:.4f} milliseconds")
```

```python
import time

def binary_search(arr, target):
    """
    Performs a binary search to find the target element in a **sorted** array.
    Returns the index of the target if found, otherwise returns -1.
    """
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# Get user input for the list
list_str = input("Enter a list of numbers separated by spaces (e.g., 10 20 30 40 50): ")
my_list = list(map(int, list_str.split()))

# Binary search requires a sorted list
my_list.sort()
print(f"Sorted List for Binary Search: {my_list}")

# Get user input for the target element
target_element = int(input("Enter the number to search for: "))

# Measure execution time
start_time = time.time()
index = binary_search(my_list, target_element)
end_time = time.time()

execution_time = (end_time - start_time) * 1000 # Convert to milliseconds

print(f"\nTarget Element: {target_element}")

if index != -1:
    print(f"Element {target_element} found at index {index} in the sorted list.")
else:
    print(f"Element {target_element} not found in the list.")

print(f"Execution Time: {execution_time:.4f} milliseconds")
```
